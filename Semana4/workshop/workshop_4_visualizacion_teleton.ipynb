## Parte 0: Setup

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
sns.set_style('whitegrid')
%matplotlib inline

# Paleta de colores Teletón
COLORES_TELETON = {
    'amarillo': '#F7C600',
    'morado_profundo': '#4B1F76',
    'morado_medio': '#7E3AA7',
    'verde': '#2ECC71',
    'rojo': '#E74C3C',
    'azul': '#1A2A6C'
}

print("✅ Setup completo")
````

## Parte 1: Cargar y Explorar Datos

````python
# Cargar datos
df = pd.read_csv('https://raw.githubusercontent.com/HesusG/diagnostico-lineas-accion/main/Semana4/datos/teleton_benefactores.csv')

print(f"Dataset: {df.shape[0]} benefactores × {df.shape[1]} variables")
df.head()
````

````python
# Variables principales
print("Columnas del dataset:")
print(df.columns.tolist())
````

## Parte 2: Principios de Visualización

### ✅ Buenas Prácticas:
1. **Usar colores con propósito** - No solo decorativos
2. **Limitar a 5-7 colores** - Evitar sobrecarga visual
3. **Etiquetar claramente** - Títulos, ejes, leyendas
4. **Mantener proporciones** - No truncar ejes Y
5. **Ordenar datos** - Mayor a menor o viceversa

### ❌ Anti-Patterns:
- Gráficos 3D innecesarios
- Muchas categorías en pie chart
- Colores que no contrastan

## Parte 3: KPIs y Scorecards

Los KPIs deben estar en la parte superior del dashboard.

````python
# Calcular KPIs principales
kpis = {
    'Satisfacción General': df['satisfaccion_general'].mean(),
    'Net Promoter Score': ((df['nps'] >= 9).sum() - (df['nps'] <= 6).sum()) / len(df) * 100,
    'Calidad Servicio': df['indice_calidad_servicio'].mean(),
    'Antigüedad Promedio': df['anos_benefactor'].mean()
}

# Mostrar KPIs
for nombre, valor in kpis.items():
    print(f"{nombre:25s}: {valor:>8.2f}")
````

````python
# Visualizar KPIs como scorecards
fig, axes = plt.subplots(1, 4, figsize=(16, 3))
fig.patch.set_facecolor(COLORES_TELETON['morado_profundo'])

kpi_values = [
    (kpis['Satisfacción General'], '/10', COLORES_TELETON['amarillo']),
    (kpis['Net Promoter Score'], '%', COLORES_TELETON['verde']),
    (kpis['Calidad Servicio'], '/5', COLORES_TELETON['morado_medio']),
    (kpis['Antigüedad Promedio'], 'años', COLORES_TELETON['azul'])
]

for ax, (nombre, valor), (kpi_val, unidad, color) in zip(axes, kpis.items(), kpi_values):
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off')
    
    # Valor principal
    ax.text(0.5, 0.5, f"{kpi_val:.1f}{unidad}", 
            ha='center', va='center', fontsize=32, fontweight='bold', color=color)
    
    # Etiqueta
    ax.text(0.5, 0.15, nombre, 
            ha='center', va='center', fontsize=12, color='white')

plt.tight_layout()
plt.show()
````

## Parte 4: Visualizaciones de Comparación

**Cuándo usar:**
- Comparar categorías
- Rankings
- Antes/después

**Tipos:** Barras horizontales, barras verticales, barras agrupadas

````python
# Top 10 giros por satisfacción
top_giros = df.groupby('giro')['satisfaccion_general'].agg(['mean', 'count']).reset_index()
top_giros = top_giros[top_giros['count'] >= 5].sort_values('mean', ascending=False).head(10)

plt.figure(figsize=(12, 6))
bars = plt.barh(top_giros['giro'], top_giros['mean'], color=COLORES_TELETON['morado_profundo'])

# Añadir línea de promedio global
media_global = df['satisfaccion_general'].mean()
plt.axvline(media_global, color=COLORES_TELETON['amarillo'], linestyle='--', linewidth=2, label=f'Promedio: {media_global:.2f}')

plt.xlabel('Satisfacción General (1-10)', fontsize=12, fontweight='bold')
plt.title('Top 10 Giros Empresariales por Satisfacción', fontsize=14, fontweight='bold', color=COLORES_TELETON['morado_profundo'])
plt.legend()
plt.tight_layout()
plt.show()
````

## Parte 5: Visualizaciones de Distribución

**Cuándo usar:**
- Entender dispersión de datos
- Identificar outliers
- Comparar distribuciones entre grupos

**Tipos:** Boxplot, violin plot, histogramas

````python
# Distribución de NPS por segmento de antigüedad
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='segmento_antiguedad', y='nps', palette='Set2')
sns.stripplot(data=df, x='segmento_antiguedad', y='nps', color='black', alpha=0.3, size=3)

plt.title('Distribución de NPS por Segmento de Antigüedad', fontsize=14, fontweight='bold', color=COLORES_TELETON['morado_profundo'])
plt.xlabel('Segmento de Antigüedad', fontsize=12)
plt.ylabel('Net Promoter Score (0-10)', fontsize=12)
plt.xticks(rotation=15, ha='right')
plt.tight_layout()
plt.show()
````

````python
# Violin plot (muestra densidad de distribución)
plt.figure(figsize=(12, 6))
sns.violinplot(data=df, x='nps_categoria', y='satisfaccion_general', palette={
    'Promotor': COLORES_TELETON['verde'],
    'Pasivo': COLORES_TELETON['amarillo'],
    'Detractor': COLORES_TELETON['rojo']
})

plt.title('Distribución de Satisfacción por Categoría NPS', fontsize=14, fontweight='bold')
plt.xlabel('Categoría NPS', fontsize=12)
plt.ylabel('Satisfacción General (1-10)', fontsize=12)
plt.tight_layout()
plt.show()
````

## Parte 6: Heatmaps y Correlaciones

**Cuándo usar:**
- Mostrar correlaciones entre variables
- Visualizar matrices de confusión
- Patrones en datos categóricos cruzados

````python
# Correlaciones entre variables de satisfacción
vars_satisfaccion = ['satisfaccion_general', 'nps', 'calidad_percibida', 'transparencia', 'indice_calidad_servicio']
correlacion = df[vars_satisfaccion].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlacion, annot=True, fmt='.2f', cmap='YlGnBu', square=True, linewidths=1, cbar_kws={'label': 'Correlación'})

plt.title('Matriz de Correlación: Variables de Satisfacción', fontsize=14, fontweight='bold', color=COLORES_TELETON['morado_profundo'])
plt.tight_layout()
plt.show()
````

## Parte 7: Dashboard Layout

### Estructura Recomendada:

```
┌─────────────────────────────────────────┐
│  HEADER: Logo + Título                  │
├───────┬───────┬───────┬────────────────┤
│ KPI 1 │ KPI 2 │ KPI 3 │ KPI 4          │ ← Scorecards
├───────┴───────┴───────┴────────────────┤
│  Gráfico Principal (50% ancho)         │
│  + Insight Textual                      │
├──────────────────┬─────────────────────┤
│ Gráfico 2        │ Gráfico 3           │
├──────────────────┴─────────────────────┤
│  Tabla de Datos + Filtros               │
└─────────────────────────────────────────┘
```

### Principios de Layout:
1. **F-Pattern**: Usuarios leen de izquierda a derecha, arriba a abajo
2. **Jerarquía**: Lo más importante arriba
3. **Agrupación**: Gráficos relacionados juntos
4. **Espacio en blanco**: No saturar

````python
# Ejemplo de dashboard compuesto
fig = plt.figure(figsize=(16, 10))
gs = fig.add_gridspec(3, 2, hspace=0.3, wspace=0.3)

# Row 1: NPS Distribution
ax1 = fig.add_subplot(gs[0, :])
segmentos_nps = df['nps_categoria'].value_counts()
colors_nps = [COLORES_TELETON['verde'], COLORES_TELETON['amarillo'], COLORES_TELETON['rojo']]
ax1.barh(segmentos_nps.index, segmentos_nps.values, color=colors_nps)
ax1.set_title('Distribución de Benefactores por Categoría NPS', fontsize=14, fontweight='bold')
ax1.set_xlabel('Cantidad de Benefactores')

# Row 2 Left: Satisfacción por Giro (Top 5)
ax2 = fig.add_subplot(gs[1, 0])
top5_giros = df.groupby('giro')['satisfaccion_general'].mean().sort_values(ascending=False).head(5)
ax2.bar(range(len(top5_giros)), top5_giros.values, color=COLORES_TELETON['morado_medio'])
ax2.set_xticks(range(len(top5_giros)))
ax2.set_xticklabels(top5_giros.index, rotation=45, ha='right')
ax2.set_title('Top 5 Giros por Satisfacción', fontsize=12, fontweight='bold')
ax2.set_ylabel('Satisfacción (1-10)')

# Row 2 Right: Distribución de Antigüedad
ax3 = fig.add_subplot(gs[1, 1])
ax3.hist(df['anos_benefactor'], bins=20, color=COLORES_TELETON['azul'], alpha=0.7, edgecolor='white')
ax3.axvline(df['anos_benefactor'].mean(), color=COLORES_TELETON['amarillo'], linestyle='--', linewidth=2)
ax3.set_title('Distribución de Antigüedad', fontsize=12, fontweight='bold')
ax3.set_xlabel('Años como Benefactor')
ax3.set_ylabel('Frecuencia')

# Row 3: Boxplot por Segmento
ax4 = fig.add_subplot(gs[2, :])
df.boxplot(column='satisfaccion_general', by='segmento_antiguedad', ax=ax4)
ax4.set_title('Satisfacción por Segmento de Antigüedad', fontsize=12, fontweight='bold')
ax4.set_xlabel('Segmento')
ax4.set_ylabel('Satisfacción (1-10)')
plt.suptitle('')  # Remover título automático de pandas

fig.suptitle('Dashboard de Satisfacción - Fundación Teletón', fontsize=16, fontweight='bold', y=0.995)
plt.show()
````

## Parte 8: Exportación para Looker Studio

````python
# Preparar datos agregados para Looker
agregaciones = {
    'kpis': pd.DataFrame([kpis]),
    'por_giro': df.groupby('giro').agg({
        'satisfaccion_general': 'mean',
        'nps': 'mean',
        'indice_calidad_servicio': 'mean',
        'giro': 'count'
    }).rename(columns={'giro': 'cantidad'}).reset_index(),
    'por_estado': df.groupby('estado').agg({
        'satisfaccion_general': 'mean',
        'estado': 'count'
    }).rename(columns={'estado': 'cantidad'}).reset_index(),
    'nps_segmentos': df['nps_categoria'].value_counts().reset_index()
}

# Exportar cada agregación
for nombre, data in agregaciones.items():
    filename = f'teleton_{nombre}_looker.csv'
    data.to_csv(filename, index=False)
    print(f"✅ Exportado: {filename}")
````
